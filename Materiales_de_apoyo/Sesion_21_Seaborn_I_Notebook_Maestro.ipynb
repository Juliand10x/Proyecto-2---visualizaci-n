# 5. Matplotlib y Seaborn: ¿qué papel cumple cada uno?

- **Matplotlib** proporciona el control fundamental para construir y personalizar gráficos en Python.
- **Seaborn** está construido sobre Matplotlib y ofrece una sintaxis de más alto nivel, especialmente cómoda para trabajar directamente con DataFrames y construir visualizaciones estadísticas.


### El mismo problema, dos formas de construir la visualización

Matplotlib y Seaborn pueden producir gráficos visualmente muy similares. La diferencia principal no está necesariamente en el resultado visual, sino en **cómo construimos la visualización**.

Comenzaremos con un histograma de la variable `EDAD`.

```python
# Histograma con Matplotlib

plt.figure(figsize=(9, 5))

plt.hist(
    edades_anios["EDAD"],
    bins=30
)

plt.title("Distribución de la edad")
plt.xlabel("Edad (años)")
plt.ylabel("Frecuencia")

plt.show()
```

### Construcción con Matplotlib

Con Matplotlib proporcionamos directamente los valores que queremos representar:

`edades_anios["EDAD"]`

Ahora construiremos el mismo histograma utilizando Seaborn.

```python
# Histograma con Seaborn

plt.figure(figsize=(9, 5))

sns.histplot(
    data=edades_anios,
    x="EDAD",
    bins=30
)

plt.title("Distribución de la edad")
plt.xlabel("Edad (años)")
plt.ylabel("Frecuencia")

plt.show()
```

### ¿Cuál es la diferencia?

Los dos gráficos comunican prácticamente la misma información.

La diferencia está en la forma de construirlos:

- Con **Matplotlib** proporcionamos directamente los valores que queremos representar.
- Con **Seaborn** indicamos el DataFrame mediante `data=` y seleccionamos las variables por su nombre mediante `x=`, `y=`, `hue=`, entre otros.

La ventaja de Seaborn se hace más evidente cuando incorporamos información estadística o nuevas variables.

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=edades_anios,
    x="EDAD",
    hue="GENERO",
    bins=30,
    kde=True
)

plt.title("Distribución de la edad según género")
plt.xlabel("Edad (años)")
plt.ylabel("Frecuencia")

plt.show()
```

En una sola instrucción de Seaborn hemos incorporado:

- `hue="GENERO"` para representar una segunda variable mediante color.
- `kde=True` para añadir la estimación de densidad.

Esto muestra una de las fortalezas de Seaborn para construir visualizaciones estadísticas directamente a partir de un DataFrame.

### Entonces, ¿por qué utilizamos ambos?

**Seaborn** facilita la construcción de visualizaciones estadísticas a partir de DataFrames.

**Matplotlib** proporciona la base gráfica y permite controlar y personalizar la figura.

Por eso los utilizaremos juntos durante la sesión:

- `sns.histplot(...)` → construye la visualización estadística.
- `plt.figure(...)` → controla la figura.
- `plt.title(...)` → define el título.
- `plt.xlabel(...)` y `plt.ylabel(...)` → personalizan los ejes.
- `plt.show()` → muestra la figura.

> **Seaborn y Matplotlib no compiten en esta sesión: cumplen funciones complementarias.**

# 7. Gráficos de barras

## Pregunta analítica

> **¿Cómo se distribuyen los registros según el género reportado?**

Antes de graficar construiremos una tabla de frecuencias con Pandas. Esto nos permite identificar con claridad **qué medida representará la longitud de cada barra**.


```python
frecuencia_genero = (
    datos["GENERO"]
    .value_counts(dropna=True)
    .rename_axis("GENERO")
    .reset_index(name="TOTAL")
)

frecuencia_genero
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=frecuencia_genero,
    x="GENERO",
    y="TOTAL"
)

plt.title("Registros según género")
plt.xlabel("Género")
plt.ylabel("Número de registros")

plt.show()
```

### ¿Qué representa la altura de las barras?

En este gráfico la altura representa el **número de registros con género informado**.

Una barra no significa automáticamente “cantidad”. Podría representar una suma, un promedio, un porcentaje u otra medida. Por eso debemos saber qué estamos graficando.


## Un ejemplo con más categorías

Ahora estudiaremos los tipos de incidente.

### Pregunta analítica

> **¿Cuáles son los tipos de incidente con mayor número de registros?**

Como existen muchas categorías, mostraremos las 10 más frecuentes y utilizaremos barras horizontales para favorecer la lectura de etiquetas extensas.


```python
# Observamos la estructura de la columna TIPO_INCIDENTE

sorted(datos["TIPO_INCIDENTE"].dropna().unique())
```

```python
# Estandarización

correcciones = {
    "ACOMPA¥AMIENTO A EVENTO": "ACOMPAÑAMIENTO A EVENTO",
    "ACOMPAÑAMIENTO A EVENTO": "ACOMPAÑAMIENTO A EVENTO",
    "CAIALT": "CAIALT - CAIDA DE ALTURA",
    "CONVULSION": "CONVULSION - CONVULSION",
    "CONVULSIÃN": "CONVULSION - CONVULSION",
    "DOLTOR": "DOLTOR - DOLOR TORACICO",
    "ELECTROCUC": "ELECTROCUC - ELECTROCUCION / RESCATE",
    "INTOX": "INTOX - INTOXICACION",
    "PATGIN": "PATGIN - PATOLOGIA GINECOBSTETRICA",
    "SINTOGASTR": "SINTOGASTR - SINTOMAS GASTROINTESTINALES"
}

datos["TIPO_INCIDENTE_LIMPIO"] = datos["TIPO_INCIDENTE"].copy()

for codigo, etiqueta in correcciones.items():
    datos.loc[
        datos["TIPO_INCIDENTE"].str.startswith(codigo, na=False),
        "TIPO_INCIDENTE_LIMPIO"
    ] = etiqueta


sorted(datos["TIPO_INCIDENTE_LIMPIO"].dropna().unique())
```

```python
# Corrección CONVULSIàN - CONVULSIàN

datos.loc[
    datos["TIPO_INCIDENTE_LIMPIO"].str.contains("CONVULSI", na=False),
    "TIPO_INCIDENTE_LIMPIO"
] = "CONVULSION - CONVULSION"


sorted(datos["TIPO_INCIDENTE_LIMPIO"].dropna().unique())
```

```python
# Top tipo incidente

top_incidentes = (
    datos["TIPO_INCIDENTE_LIMPIO"]
    .value_counts()
    .head(10)
    .sort_values(ascending=True)
    .rename_axis("TIPO_INCIDENTE")
    .reset_index(name="TOTAL")
)

top_incidentes
```

```python
# Gráfico de barras

plt.figure(figsize=(11, 7))

sns.barplot(
    data=top_incidentes,
    x="TOTAL",
    y="TIPO_INCIDENTE"
)

plt.title("Diez tipos de incidente con mayor número de registros")
plt.xlabel("Número de registros")
plt.ylabel("")

plt.show()
```

### Observa antes de interpretar

La base puede contener diferencias de escritura o codificación en algunas categorías. Si encuentras dos etiquetas que parecen representar el mismo concepto, **no las unas automáticamente**.

Trabajaremos tres familias:

1. **Cualitativas:** categorías diferentes sin orden natural.
2. **Secuenciales:** magnitudes ordenadas de menor a mayor.
3. **Divergentes:** valores que se alejan en dos direcciones respecto a un punto de referencia.


## 9.1 Paleta cualitativa

Es apropiada cuando queremos diferenciar categorías que no poseen un orden natural.

```python
sns.color_palette("Set2")
```

## 9.2 Paleta secuencial

Una paleta secuencial comunica una progresión de menor a mayor.

Para demostrarla ordenaremos los diez tipos de incidente según su frecuencia y utilizaremos la intensidad del color como apoyo a la magnitud.


```python
top_incidentes_desc = (
    datos["TIPO_INCIDENTE_LIMPIO"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_incidentes_desc.columns = ["TIPO_INCIDENTE", "TOTAL"]

plt.figure(figsize=(11, 7))

sns.barplot(
    data=top_incidentes_desc,
    x="TOTAL",
    y="TIPO_INCIDENTE",
    hue="TOTAL",
    palette="Blues",
    legend=False
)

plt.title("Tipos de incidente con mayor número de registros")
plt.xlabel("Número de registros")
plt.ylabel("Tipo de incidente")

plt.show()
```

### Reflexiona

La longitud de las barras **ya comunica la magnitud**.

# 10. Gráfico de dispersión

Un gráfico de dispersión estudia la relación entre **dos variables cuantitativas**.

En la base original no conviene elegir dos columnas al azar solo para producir un `scatterplot`. Construiremos un resumen por código de localidad y plantearemos una pregunta exploratoria:

> **¿Existe alguna relación visible entre la edad promedio registrada y el número de registros por localidad?**

Esta visualización es **exploratoria**. No permite concluir causalidad.


```python
resumen_localidad = (
    edades_anios
    .groupby("CODIGO_LOCALIDAD")
    .agg(
        TOTAL_REGISTROS=("NUMERO_INCIDENTE", "size"),
        EDAD_PROMEDIO=("EDAD", "mean")
    )
    .reset_index()
)

resumen_localidad.head()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=resumen_localidad,
    x="EDAD_PROMEDIO",
    y="TOTAL_REGISTROS",
    s=80
)

plt.title("Edad promedio y número de registros por localidad")
plt.xlabel("Edad promedio (años)")
plt.ylabel("Número de registros")

plt.show()
```
